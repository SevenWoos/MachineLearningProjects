# Regexes

* Regexes = pattern matching of strings in Python.
  - EX: password checkers, phone numbers, emails, etc.

* regex101
  - https://regex101.com/

* Any number of characters:
  - .

* O or more
  - *

* Whitespace
  - \s

* Not a, b, c
  - [^abc]

* Not
  - [^]

* Start of string
  - ^

* End of string
  - $

* EX:
  - start with letters 'ab'

  - any characters in-between(no white space)

  - end with letters 'cd'

  - '^ab[^\s]*cd$'

```python
import re
```

```python
regexp = re.compile(r"^ab[^\s]*cd$")
```

```python
phrases = ["abcd", "xxx", "abxxcd", "ab cd"]

matches = []
for phrase in phrases:
  if re.match(regexp, phrase):
    matches.append(phrase)
print(matches)
```

### Search

* What if we wanted to find our matches in a string?
  - EX: "aaa abxxxcd cccc" CONTAINS a match in the middle.

```python
regexp = re.compile(r"ab[^\s]*cd")
phrases = ["abcd", "xxx", "aaa abxxcd ccc", "ab cd"]

matches = []
for phrase in phrases:
  if re.search(regexp, phrase):
    matches.append(phrase)
print(matches)
```

```python
regexp = re.compile(r"read|story|book")

# ""
phrases = ["I liked that story", "i like that book", "this hat is nice"]

matches = []
for phrase in phrases:
  if re.search(regexp, phrase):
    matches.append(phrase)
print(matches)
```

```python
regexp = re.compile(r"read|story|book")

# The or(|) has word boundaries, so "read" in "thread" will match.
# "story" in "history" will match
phrases = ["I liked that history", "i like that book", "this hat is nice", "the car threaded up the hill"]

matches = []
for phrase in phrases:
  if re.search(regexp, phrase):
    matches.append(phrase)
print(matches)
```

```python
# Adding WORD BOUNDARIES with '\b'
regexp = re.compile(r"\bread\b|\bstory\b|\bbook\b")

phrases = ["I liked that story", "i like that book", "this hat is nice", "the car threaded up the hill"]

matches = []
for phrase in phrases:
  if re.search(regexp, phrase):
    matches.append(phrase)
print(matches)
```
