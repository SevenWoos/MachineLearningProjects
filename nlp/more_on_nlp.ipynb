```python
from textblob import TextBlob
```

```python
# Create a TextBlob object
phrase = "this is an example"

tb_phrase = TextBlob(phrase)

tb_phrase.correct()
```

```python
# Spell Correct
phrase = "thiiis is an examplee"

tb_phrase = TextBlob(phrase)

tb_phrase.correct()
```

```python
# Part of Speech tagging
phrase = "i read the book"

tb_phrase = TextBlob(phrase)

tb_phrase.tags
```

```python
# Sentiment
phrase = "i read the book"

tb_phrase = TextBlob(phrase)

tb_phrase.sentiment
```

```python
# Positive Sentiment
# Higher polarity
phrase = "the book was great"

tb_phrase = TextBlob(phrase)

tb_phrase.sentiment
```

```python
# Negative Sentiment
# Negative polarity

phrase = "the book sucked it was so bad"

tb_phrase = TextBlob(phrase)
 
tb_phrase.sentiment
```
