# Stemming and Lemmatization

* Stemming and Lemmatization = technique to normalize text.

* EX: When we trained the model with the word "book", it did NOT know the word "books".
  - To us, it is straightforward.

  - Stemming and Lemmatization can take "books" and REDUCE it to down to a more cCANONICAL FORM OF "book".

* EX: Stemming
  - "reading" --> "read"

  - "books" --> "book"

* EX (Stemming Follows an ALGORITHM, SO IT DOES NOT GUARANTEE AN ACTUAL TRUE WORD):
  - "stories" --> "stori"

* EX: Lemmatization uses a DICTIONARY, and makes sure the outputs are actual words.
  - "stories" --> "story"

```python
# Use NLTK Library
import nltk

nltk.download('wordnet')
nltk.download('stopwords')
nltk.download('punkt')
```

### Stemming

```python
# Start with a stemmer.
# Use tokenize to break sentences down into words.
from nltk.tokenize import word_tokenize

# Stemmer
from nltk.stem import PorterStemmer
```

```python
stemmer = PorterStemmer()

phrase = "reading the books"

# Tokenize the phrase.
words = word_tokenize(phrase)

# Stem the words.
stemmed_words = []
for word in words:
  stemmed_words.append(stemmer.stem(word))
" ".join(stemmed_words)
```

```python
stemmer = PorterStemmer()

# "stories" gets us "stori"
phrase = "stories"

# Tokenize the phrase.
words = word_tokenize(phrase)

# Stem the words.
stemmed_words = []
for word in words:
  stemmed_words.append(stemmer.stem(word))
" ".join(stemmed_words)
```

```python
# Adding punctuation at the end. It counts as a word.
stemmer = PorterStemmer()

phrase = "reading the books."

# Tokenize the phrase.
words = word_tokenize(phrase)

# Stem the words.
stemmed_words = []
for word in words:
  stemmed_words.append(stemmer.stem(word))
" ".join(stemmed_words)
```

### Lemmatization

* Expects the part of speech.

* By default, it says each token is a noun.
  - EX: "books" --> "book"

  - EX: "reading" NOT converted to "read"

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()

phrase = "reading the books"
# Tokenize the phrase.
words = word_tokenize(phrase)

lemmatized_words = []
for word in words:
  lemmatized_words.append(lemmatizer.lemmatize(word))
" ".join(lemmatized_words)
```

```python
# By default, it says each token is a noun.
# EX: "books" --> "book"
# EX: "reading" NOT converted to "read"
lemmatizer = WordNetLemmatizer()

phrase = "reading the books"
# Tokenize the phrase.
words = word_tokenize(phrase)

lemmatized_words = []
for word in words:
  lemmatized_words.append(lemmatizer.lemmatize(word, pos='v'))
" ".join(lemmatized_words)
```
