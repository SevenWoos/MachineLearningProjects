* Recurrent Neural Networks can handle different amounts of input data and use a FEEDBACK LOOP.
  - Great for sequential input as it can feed information from the last network to the next network.

  - We will get a word embedding from this.



* RNN Drawbacks.
  - LONGER dependencies don't always perform well.

  - "I need to go to the bank today so that I can make a deposit, I hope it's not closed."

    - By the time, we get to the end "closed", a word like "bank" is alr pretty FAR AWAY.

    - It might NOT be clear to the network that the thing we're talking about being "closed" is the bank since it is so FAR AWAY.

* RNN Performance Issues
  - The SEQUENTIAL nature of RNNs(we have to feed in input ONE AT A TIME), makes it TOUGH to PARALLELIZE the training.

  - Can't effectively utilized modern GPUs.
