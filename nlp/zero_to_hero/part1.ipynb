# NLP Zero To Hero Part 1: Tokenization

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
sentences = [
  'I love my dog', 
  'I love my cat'
]
```

```python
# Tokenizer object, where "num_words" = max number of words to keep.
tokenizer = Tokenizer(num_words = 100)
# Fit the tokenizer based on instance.
tokenizer.fit_on_texts(sentences)
# "word_index" contains the tokenizer's full list of words
word_index = tokenizer.word_index
print(word_index)
```

```python
# Adding an exclamation point. 
# Tokenizer is smart enough to spot this, and not create a NEW token. Leaves it as "dog".
sentences = [
  'I love my dog', 
  'I love my cat', 
  'You love my dog!'
]
```

```python
tokenizer = Tokenizer(num_words = 100)
tokenizer.fit_on_texts(sentences)
word_index = tokenizer.word_index
print(word_index)
```
