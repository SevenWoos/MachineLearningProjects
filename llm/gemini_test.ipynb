```python
# import google.generativeai as genai 
import os
from dotenv import load_dotenv
load_dotenv()  # read .env file
```

```python
# Create a client object once that will read the API key and handle all requests through it.
client = genai.Client(api_key = os.environ["GEMINI_API_KEY"])
```

```python
# Test our key actually loaded in.
# print(os.environ.get("GEMINI_API_KEY")) 
```

```python
# Create reusable function to call Gemini API.
def ask_gemini(prompt: str, temperature: float = 0.7, max_tokens: int = 300) -> str:
  """
  Send a prompt to Gemini and return the generated text.

  prompt: the text we're asking the model to respond to
  temperature: 0 = more focused/deterministic, 1 = more varied/creative
  max_tokens: caps how long the response can be
  """
  
  # generate_content now lives on client.models, and takes the model
  # name as an explicit argument each call.
  
  # Gemini 3.6 Flash ships with thinking mode on by default, so thinking tokens COUNT against the max_output_tokens budget alongside the actual visible response. 
  # Some of that 300-token budget is  consumed by internal reasoning before the model even starts writing the returned answer.
  response = client.models.generate_content(
    model="gemini-3.6-flash", 
    contents=prompt, 
    config=genai.types.GenerateContentConfig(
      temperature=temperature, 
      max_output_tokens=max_tokens, 
      
      # We can disable thinking(if we don't need deep reasoning for a simple explanation, which saves tokens and latency)
      # "Thinking" refers to the model's internal reasoning process — extra generation steps where it works through a problem step-by-step before producing a final answer (similar in spirit to chain-of-thought), using nothing but its own trained knowledge.
      # Could also raise max_tokens limit in parameter
      thinking_config=genai.types.ThinkingConfig(thinking_budget=0),
    ), 
  )
  
  return response.text
  
```

```python
result = ask_gemini("Explain what a transformer model is, in two sentences.")
print(result)
```

```python
result = ask_gemini("What are Transformers from the Hssbro toy brand? Describe in two sentences?")
print(result)
```
