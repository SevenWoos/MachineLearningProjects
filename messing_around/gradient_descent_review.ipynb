# Gradient Descent Review

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
# Generating some simple data for a linear relationship
np.random.seed(0)
```

```python
# np.random.rand(100, 1) generates random (100 rows x 1 column) matrix of data points.
# Each number = randomly generated from a uniform distribution -> all #s between 0 & 1 have equal chance of being selected.
# Multiplying by 2 scales the range from (0-1) to (0-2)
X = 2 * np.random.rand(100, 1)
```

```python
X[0:5]
```

```python
# 3*X multiplies each element of X by 3. X contains values between 0 & 2.
# Gives a linear relationship where the slope = 3. 
# For every 1 unit increase in X, y will increase by 3 units (in a perfectly linear model).

# 4 = constant term (intercept). Shifts line upwards by 4 units. 
# Equation (y = 4 + 3 * X) represents a straight line with:...
# Slope = 3 (how steep line is), Intercept = 4 (where line crosses y-axis).

# ADDS RANDOM NOISE to data. 
# np.random.randn(100, 1) generates 100 random numbers from a normal distribution (mean = 0, SD = 1). 
# Noise added to make y values less perfect resemble real-world data(usually has some variation or noise).
y = 4 + 3 * X + np.random.randn(100, 1)
```

```python
y[0:5]
```

```python
# Gradient Descent Parameters

# Step size
learning_rate = 0.1
# Number of steps
iterations = 1000

# Initial values for slope (m) and intercept (b)
m = 0
b = 0

# Number of data points
n = len(X)
```

```python
n
```

```python
# Gradient Descent algorithm
for i in range(iterations):
    # Current predicted values for y.
    y_pred = m*X + b
    
    # Mean squared error is the Cost/loss function
    cost = np.sum((y - y_pred)**2) / n
    
    # Gradient optimizing slope
    slope_gradient = np.sum(X * (y - y_pred)) * (-2/n)
    
    # Gradient optimizing intercept
    intercept_gradient = np.sum(y - y_pred) * (-2/n)
    
    # Update slope
    m = m - (slope_gradient*learning_rate)
    
    # Update intercept
    b = b - (intercept_gradient*learning_rate)

print(f"Optimized slope (m): {m}")
print(f"Optimized intercept (b): {b}")
```

```python
# Plotting the result
plt.scatter(X, y)
plt.plot(X, m * X + b, color='red')  # Line after training
plt.title("Line Fitted using Gradient Descent")
plt.xlabel("X")
plt.ylabel("y")
plt.show()
```
